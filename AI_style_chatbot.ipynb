```python

print("===========AI CHATBOT==========")
print("Hello! I am your personal chatbot.")
print("\nI am here to chat with you.")
name=input("\nwhat is your name? ")
print("\nNice to meet you,",name,"!\nI am happy to chat with you.")
chat_history=[]
while True:
    data= input("ask anything: ")
    print("you:",data)
    chat_history.append(data)

    if data.lower() == "hello":
        print("chatbot: Hello! how can i help you?")
    elif data.lower() == "hi":
        print("chatbot: Hey! how are you?")
    elif data.lower() == "good morning":
        print("chatbot: good morning! have a great day")
    elif data.lower() == "hey":
        print("chatbot: Hey there! nice to meet you.")
    elif data.lower() == "do you know my name?":
        print("chatbot:yes,your name is",name)
    elif data.lower() == "what is your name?" or data.lower() == "who are you":
        print("chatbot: I am your AI chatbot.")
    elif data.lower() == "who am i?":
        print("chatbot: you are",name)
    elif data.lower() == "thank you":
        print("chatbot: you are always welcome!",name,"🥰")
    elif data.lower() == "what can you do?" or data.lower() == "what do you do?":
        print("chatbot: I can chat,answer questions  and entertain you!")
    elif data.lower() == "where do you live?":
        print("chatbot: I live in the digital world!")
    elif data.lower() == "how old are you?":
        print("chatbot: I am an AI.I don't have an age")
    elif data.lower() == "i am happy":
        print("chatbot: That's wonderful! keep smiling 😊")
    elif data.lower() == "i am sad":
        print("chatbot: Oh no! I'm here if you want to talk.❤️")
    elif data.lower() == "i am angry":
        print("chatbot: Take a deep breath.everything will be okay.")
    elif data.lower() == "i am bored":
        print("chatbot: Let's play a game")
    else:
        print("chatbot: sorry,I don't understand that yet.")

    choice=input("would you like to chat further more? ")
    if choice.lower() == "no":
        print("thanks for chatting! ")
        print("Your chat history:", chat_history)
        break
    elif choice.lower() == "yes":
        print("lets continue chatting!")
```
